```python
customers_df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv("/Volumes/ecommerce/project/bronze/customers.csv")
)

display(customers_df)
customers_df.printSchema()
```

```python
from pyspark.sql.functions import *

customers_bronze_df = (
    customers_df
    .withColumn("load_timestamp", current_timestamp())
    .withColumn("source_file_name",lit('customers_raw'))
)
display(customers_bronze_df)
```

```python
(
    customers_bronze_df
    .write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("customers_raw")
)
display(customers_bronze_df)
```
